```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import os
import io
from PIL import Image

def draw_boxes_on_image(image_path, label_path):
    img = Image.open(image_path).convert("RGB")
    img_width, img_height = img.size

    fig, ax = plt.subplots(1)
    ax.imshow(img)

    if os.path.exists(label_path):
        with open(label_path, 'r') as f:
            for line in f.readlines():
                parts = line.strip().split()
                if len(parts) == 5:
                    class_id, x_center, y_center, width, height = map(float, parts)
                    x1 = (x_center - width / 2) * img_width
                    y1 = (y_center - height / 2) * img_height
                    box_width = width * img_width
                    box_height = height * img_height
                    rect = patches.Rectangle((x1, y1), box_width, box_height, linewidth=1, edgecolor='r', facecolor='none')
                    ax.add_patch(rect)


    ax.set_axis_off()
    plt.tight_layout()
    buf = io.BytesIO()
    fig.savefig(buf, format='png')
    buf.seek(0)
    annotated_img = Image.open(buf).convert("RGB")
    plt.close(fig)

    return annotated_img

for img_path, label_path in image_label_pairs[:5]:
    annotated_img = draw_boxes_on_image(img_path, label_path)
    fig, ax = plt.subplots(figsize=(6, 6))
    ax.imshow(annotated_img)
    ax.set_title(os.path.basename(img_path))
    ax.axis('off')
    plt.show()
```

```python
# using test dataset and best models to count the number of stinkbugs 

import os
import csv
from ultralytics import YOLO
import cv2


model_dir = "saved_models_of_1008_images"          
test_images_dir = "./stinkbug_image_data_third_file/test/images"    
output_root = "model_predictions"  
metrics_csv = "object_counts.csv"   

os.makedirs(output_root, exist_ok=True)

# Collect all model files
model_files = [f for f in os.listdir(model_dir) if f.endswith(".pt")]
image_files = [f for f in os.listdir(test_images_dir) if f.lower().endswith((".jpg", ".png", ".jpeg"))]

# Open CSV to log object counts
with open(metrics_csv, mode="w", newline="") as csvfile:
    writer = csv.writer(csvfile)
    header = ["image"] + [model_file.replace("_best.pt", "") for model_file in model_files]
    writer.writerow(header)

    for image_file in image_files:
        image_path = os.path.join(test_images_dir, image_file)
        row = [image_file]

        for model_file in model_files:
            model_name = model_file.replace("_best.pt", "")
            model_path = os.path.join(model_dir, model_file)
            model = YOLO(model_path)

            # Predict
            results = model.predict(source=image_path, conf=0.25, save=False, imgsz=640)

            # Count number of detections
            preds = results[0].boxes
            num_objects = len(preds)
            row.append(num_objects)

            # Save visualization
            result_img = results[0].plot() 

            # Create output dir for this model
            model_output_dir = os.path.join(output_root, model_name)
            os.makedirs(model_output_dir, exist_ok=True)

            # Save image
            save_path = os.path.join(model_output_dir, image_file)
            cv2.imwrite(save_path, result_img)

            print(f"Saved: {save_path} | Count: {num_objects} objects")

        writer.writerow(row)

print(f"\n All done! Object counts saved to: {metrics_csv}")
```

```python
def get_counts_from_labels(labels_dir, csv_output_path):
    import os
    import csv

    counts = {}
    for label_file in os.listdir(labels_dir):
        if label_file.endswith(".txt"):
            with open(os.path.join(labels_dir, label_file)) as f:
                counts[label_file.replace('.txt', '.jpg')] = len(f.readlines())
    
    with open(csv_output_path, 'w', newline='') as csvfile:
        writer = csv.writer(csvfile)
        writer.writerow(['image', 'true_count'])
        for img, count in counts.items():
            writer.writerow([img, count])
    
    print(f"Saved ground truth counts to {csv_output_path}")

# Usage:
get_counts_from_labels("./stinkbug_image_data_third_file/train/labels", "train_ground_truth_counts.csv")
get_counts_from_labels("./stinkbug_image_data_third_file/valid/labels", "val_ground_truth_counts.csv")
get_counts_from_labels("./stinkbug_image_data_third_file/test/labels", "test_ground_truth_counts.csv")
```

```python
import numpy as np

def calculate_metrics(true_counts, pred_counts):
    N = len(true_counts)
    min_vals = np.minimum(true_counts, pred_counts)
    max_vals = np.maximum(true_counts, pred_counts)
    aca = np.sum(min_vals / max_vals) / N

    detection_rate = np.sum(pred_counts >= true_counts) / N
    failed_detection_rate = 1 - detection_rate

    return aca, detection_rate, failed_detection_rate

results = []
for model in pred_counts.columns[1:]: 
    aca, dr, fdr = calculate_metrics(df['true_count'], df[model])
    results.append({
        'model': model,
        'Average Counting Accuracy': aca,
        'Detection Rate': dr,
        'Failed Detection Rate': fdr
    })

metrics_df = pd.DataFrame(results)
print(metrics_df)
metrics_df.to_csv("model_counting_metrics.csv", index=False)
```

```python
import pandas as pd
import numpy as np

def calculate_counting_metrics(true_counts, pred_counts):
    """Compute ACA, MAE, DR, and FDR between predicted and true counts."""
    true_counts = np.array(true_counts)
    pred_counts = np.array(pred_counts)

    # Avoid divide by zero
    max_vals = np.maximum(true_counts, pred_counts)
    aca = np.mean(np.where(max_vals == 0, 1, np.minimum(true_counts, pred_counts) / max_vals))
    mae = np.mean(np.abs(true_counts - pred_counts))
    dr = np.mean(pred_counts >= true_counts)
    fdr = 1 - dr

    return round(aca, 3), round(mae, 3), round(dr, 3), round(fdr, 3)

# === Load your data ===

ground_truth_csv = "test_ground_truth_counts.csv"   # columns: image,true_count
predictions_csv = "object_counts.csv"            # columns: image, yolov8n, yolov8s, ...

# Load
gt_df = pd.read_csv(ground_truth_csv)
pred_df = pd.read_csv(predictions_csv)

# Merge on image
df = pd.merge(gt_df, pred_df, on="image")

# === Evaluate models ===

metrics = []
for model in pred_df.columns[1:]: 
    aca, mae, dr, fdr = calculate_counting_metrics(df["true_count"], df[model])
    metrics.append({
        "Model": model,
        "Average Counting Accuracy": aca,
        "Mean Absolute Error": mae,
        "Detection Rate": dr,
        "Failed Detection Rate": fdr
    })

metrics_df = pd.DataFrame(metrics)
metrics_df.to_csv("model_counting_metrics.csv", index=False)
print(metrics_df)
```

```python
# creating confusion matrix plots for test and val splits

import os
from pathlib import Path
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import precision_recall_curve, confusion_matrix, ConfusionMatrixDisplay
from tqdm.auto import tqdm
import matplotlib.patches as patches

# try import ultralytics
try:
    from ultralytics import YOLO
except Exception as e:
    raise RuntimeError("ultralytics is required for this cell. Install with: pip install ultralytics")

ROOT = Path('.')
models_dir = ROOT / 'saved_models_of_1008_images'
data_root = ROOT / 'stinkbug_image_data_third_file'
output_base = ROOT / 'evaluation_outputs_of_1008'

# configuration
IOU_THRESH = 0.5
CONF_THRESH = 0.25
IMAGE_EXTS = ('.jpg', '.jpeg', '.png')

# helpers

def xywh_to_xyxy(xywh, img_w, img_h):
    x_c, y_c, w, h = xywh
    x1 = (x_c - w / 2.0) * img_w
    y1 = (y_c - h / 2.0) * img_h
    x2 = (x_c + w / 2.0) * img_w
    y2 = (y_c + h / 2.0) * img_h
    return [float(x1), float(y1), float(x2), float(y2)]


def iou(boxA, boxB):
    # boxes: [x1,y1,x2,y2]
    xA = max(boxA[0], boxB[0])
    yA = max(boxA[1], boxB[1])
    xB = min(boxA[2], boxB[2])
    yB = min(boxA[3], boxB[3])
    interW = max(0, xB - xA)
    interH = max(0, yB - yA)
    interArea = interW * interH
    boxAArea = max(0, boxA[2] - boxA[0]) * max(0, boxA[3] - boxA[1])
    boxBArea = max(0, boxB[2] - boxB[0]) * max(0, boxB[3] - boxB[1])
    denom = boxAArea + boxBArea - interArea
    if denom <= 0:
        return 0.0
    return interArea / denom


def read_gt_labels(label_path, img_w, img_h):
    boxes = []
    if not label_path.exists():
        return boxes
    with open(label_path, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if not parts:
                continue
            cls = int(parts[0])
            xywh = list(map(float, parts[1:5]))
            boxes.append({'cls': cls, 'box': xywh_to_xyxy(xywh, img_w, img_h), 'matched': False})
    return boxes


def ensure_dir(p: Path):
    p.mkdir(parents=True, exist_ok=True)


def plot_and_save_confusion(cm_counts, out_dir: Path, model_name: str, split: str):
    """Custom confusion plotting: diagonal cells filled blue with white text, off-diagonals white with black text.
    Rows are PREDICTED classes, columns are ACTUAL classes (this is the detection convention used here).
    Saved figures will have:
      - Y-axis: label 'Predicted' (vertical), tick labels vertical
      - X-axis: label 'Actual' (horizontal), tick labels horizontal
      - Colorbar label (if shown) vertical
    This variant shows ONLY raw counts (no percentage annotations).
    """
    ensure_dir(out_dir)

    # Explicit mapping for this binary problem (predicted row, actual column)
    # row 0 = predicted: target, row 1 = predicted: background
    # col 0 = actual: target, col 1 = actual: background
    # Replace with your class names as needed
    pred_names = ['prediction: Redbanded stink bug', 'prediction: background']
    actual_names = ['actual: Redbanded stink bug', 'actual: background']

    # --- Raw counts panel ---
    fig, ax = plt.subplots(figsize=(6,5), dpi=300)
    ax.set_xlim(0,2); ax.set_ylim(0,2)
    ax.invert_yaxis()

    # Draw 2x2 cells with diagonal blue
    diag_color = '#1f77b4'  # blue
    off_color = '#ffffff'   # white
    edge_color = 'black'

    for i in range(2):
        for j in range(2):
            face = diag_color if i == j else off_color
            rect = patches.Rectangle((j, i), 1, 1, facecolor=face, edgecolor=edge_color, linewidth=2)
            ax.add_patch(rect)
            # annotation: count
            try:
                count = int(cm_counts[i, j])
            except Exception:
                count = 0
            txt_color = 'white' if i == j else 'black'
            ax.text(j + 0.5, i + 0.5, f"{count}", ha='center', va='center', color=txt_color, fontsize=28, fontweight='bold')

    # Set ticks: columns are ACTUAL, rows are PREDICTED
    ax.set_xticks([0.5, 1.5])
    # x tick text horizontally (rotation=0)
    ax.set_xticklabels([n.replace('actual: ', '') for n in actual_names], fontsize=14, rotation=0, ha='center', va='top')
    ax.set_yticks([0.5, 1.5])
    # y tick text vertically (rotation=90)
    ax.set_yticklabels([n.replace('prediction: ', '') for n in pred_names], fontsize=14, rotation=90, ha='center', va='center')

    # Axis labels with requested orientations
    ax.set_xlabel('Actual', fontsize=14, rotation=0, labelpad=12)
    ax.set_ylabel('Predicted', fontsize=14, rotation=90, labelpad=12)

    ax.set_title(f'Confusion matrix', fontsize=16)
    plt.tight_layout()
    fig_path_png = out_dir / 'confusion_matrix_counts.png'
    fig_path_pdf = out_dir / 'confusion_matrix_counts.pdf'
    fig.savefig(str(fig_path_png))
    fig.savefig(str(fig_path_pdf))
    plt.close(fig)

    # --- Second panel: counts-only (larger text) ---
    fig, ax = plt.subplots(figsize=(6,5), dpi=300)
    ax.set_xlim(0,2); ax.set_ylim(0,2)
    ax.invert_yaxis()

    for i in range(2):
        for j in range(2):
            face = diag_color if i == j else off_color
            rect = patches.Rectangle((j, i), 1, 1, facecolor=face, edgecolor=edge_color, linewidth=2)
            ax.add_patch(rect)
            try:
                count = int(cm_counts[i, j])
            except Exception:
                count = 0
            txt_color = 'white' if i == j else 'black'
            # annotate count
            ax.text(j + 0.5, i + 0.5, f"{count}", ha='center', va='center', color=txt_color, fontsize=32, fontweight='bold')

    ax.set_xticks([0.5, 1.5])
    ax.set_xticklabels([n.replace('actual: ', '') for n in actual_names], fontsize=14, rotation=0, ha='center', va='top')
    ax.set_yticks([0.5, 1.5])
    ax.set_yticklabels([n.replace('prediction: ', '') for n in pred_names], fontsize=14, rotation=90, ha='center', va='center')

    ax.set_xlabel('Actual', fontsize=14, rotation=0, labelpad=12)
    ax.set_ylabel('Predicted', fontsize=14, rotation=90, labelpad=12)

    ax.set_title(f'Confusion matrix (counts) ({model_name} - {split})', fontsize=16)
    plt.tight_layout()
    fig_path_png = out_dir / 'confusion_matrix_normalized_row_annotated.png'
    fig_path_pdf = out_dir / 'confusion_matrix_normalized_row_annotated.pdf'
    fig.savefig(str(fig_path_png))
    fig.savefig(str(fig_path_pdf))
    plt.close(fig)


def evaluate_model_on_split(model_path: Path, split: str, max_images=None):
    # split: 'test' or 'valid' (data uses 'val' naming in data.yaml; folder is 'valid' in this project)
    split_folder = 'test' if split == 'test' else 'valid'
    imgs_dir = data_root / split_folder / 'images'
    labels_dir = data_root / split_folder / 'labels'

    if not imgs_dir.exists():
        print(f"Images directory for split '{split}' not found: {imgs_dir}")
        return

    model = YOLO(str(model_path))
    images = sorted([p for p in imgs_dir.iterdir() if p.suffix.lower() in IMAGE_EXTS])
    if max_images:
        images = images[:max_images]

    all_confidences = []
    all_labels = []
    TP = FP = FN = 0
    detection_counts = []

    for img_p in tqdm(images, desc=f"{model_path.name} - {split}"):
        img = Image.open(img_p)
        img_w, img_h = img.size
        label_file = labels_dir / (img_p.stem + '.txt')
        gt_boxes = read_gt_labels(label_file, img_w, img_h)

        # run inference - get predictions as boxes with confidences
        results = model.predict(source=str(img_p), conf=CONF_THRESH, imgsz=640, verbose=False)
        res = results[0]
        preds = []
        if hasattr(res, 'boxes') and res.boxes is not None:
            for box in res.boxes:
                # box.xyxy, box.conf, box.cls
                xyxy = box.xyxy.cpu().numpy().tolist() if hasattr(box.xyxy, 'cpu') else box.xyxy.tolist()
                if isinstance(xyxy, list) and len(xyxy) == 1 and isinstance(xyxy[0], (list, tuple)):
                    xyxy = xyxy[0]
                if not (isinstance(xyxy, (list, tuple)) and len(xyxy) == 4):
                    continue
                xyxy = [float(x) for x in xyxy]
                conf = float(box.conf.cpu().numpy()) if hasattr(box.conf, 'cpu') else float(box.conf)
                cls = int(box.cls.cpu().numpy()) if hasattr(box.cls, 'cpu') else int(box.cls)
                preds.append({'cls': cls, 'box': xyxy, 'conf': conf})

        preds = sorted(preds, key=lambda x: x['conf'], reverse=True)
        detection_counts.append(len(preds))

        for p in preds:
            matched = False
            for gt in gt_boxes:
                if gt['matched']:
                    continue
                if gt['cls'] != p['cls']:
                    continue
                if iou(p['box'], gt['box']) >= IOU_THRESH:
                    TP += 1
                    gt['matched'] = True
                    matched = True
                    all_confidences.append(p['conf'])
                    all_labels.append(1)
                    break
            if not matched:
                FP += 1
                all_confidences.append(p['conf'])
                all_labels.append(0)

        for gt in gt_boxes:
            if not gt['matched']:
                FN += 1

    # confusion matrix: [[TP, FN],[FP, TN]] ; TN set to 0 in detection context
    cm = np.array([[TP, FN], [FP, 0]])

    # output dir
    out_dir = output_base / model_path.stem / split
    ensure_dir(out_dir)

    # Save improved confusion matrices using our custom helper (counts-only)
    try:
        plot_and_save_confusion(cm, out_dir, model_path.stem, split)
    except Exception as e:
        print('Failed to plot with custom helper, falling back to matplotlib-only plots:', e)
        plt.figure(figsize=(5,4))
        plt.imshow(cm, cmap='Blues')
        for i in range(cm.shape[0]):
            for j in range(cm.shape[1]):
                plt.text(j, i, int(cm[i, j]), ha='center', va='center', color='black')
        # make y-axis tick labels vertical in fallback as well
        ax = plt.gca()
        ax.set_xticks([0, 1])
        ax.set_xticklabels(['actual: Redbanded stink bug', 'actual: background'], rotation=0)
        ax.set_yticks([0, 1])
        ax.set_yticklabels(['prediction: Redbanded stink bug', 'prediction: background'], rotation=90, ha='center', va='center')
        plt.xlabel('Actual')
        plt.ylabel('Predicted')
        plt.title(f'Confusion matrix')
        plt.tight_layout()
        plt.savefig(out_dir / 'confusion_matrix_counts.png')
        plt.close()

    # PR curve
    if len(all_confidences) > 0:
        precisions, recalls, thresholds = precision_recall_curve(all_labels, all_confidences)
        plt.figure()
        plt.plot(recalls, precisions, marker='.')
        plt.xlabel('Recall')
        plt.ylabel('Precision')
        plt.title(f'Precision-Recall curve ({model_path.stem} - {split})')
        plt.grid(True)
        plt.tight_layout()
        plt.savefig(out_dir / 'pr_curve.png')
        plt.close()

        thr_plot = np.concatenate(([0.0], thresholds))
        prec_plot = precisions
        rec_plot = recalls
        plt.figure()
        plt.plot(thr_plot, prec_plot, label='Precision')
        plt.plot(thr_plot, rec_plot, label='Recall')
        plt.xlabel('Confidence threshold (approx)')
        plt.ylabel('Score')
        plt.title(f'Precision/Recall vs threshold ({model_path.stem} - {split})')
        plt.legend()
        plt.grid(True)
        plt.tight_layout()
        plt.savefig(out_dir / 'precision_recall_vs_threshold.png')
        plt.close()

    # detection count histogram
    plt.figure()
    plt.hist(detection_counts, bins=range(0, max(detection_counts) + 2 if detection_counts else 2), color='tab:blue', alpha=0.8)
    plt.xlabel('Number of detections (per image)')
    plt.ylabel('Image count')
    plt.title(f'Detection counts histogram ({model_path.stem} - {split})')
    plt.tight_layout()
    plt.savefig(out_dir / 'detection_counts_histogram.png')
    plt.close()

    # write summary
    total_images = len(images)
    precision_approx = TP / (TP + FP) if (TP + FP) > 0 else 0.0
    recall_approx = TP / (TP + FN) if (TP + FN) > 0 else 0.0
    with open(out_dir / 'summary.txt', 'w') as f:
        f.write(f'Model: {model_path.name}\n')
        f.write(f'Split: {split}\n')
        f.write(f'Images evaluated: {total_images}\n')
        f.write(f'TP: {TP}\n')
        f.write(f'FP: {FP}\n')
        f.write(f'FN: {FN}\n')
        f.write(f'Precision (approx): {precision_approx:.4f}\n')
        f.write(f'Recall (approx): {recall_approx:.4f}\n')

    print(f'Evaluation outputs saved to: {out_dir} (images: {total_images})')


# Main orchestration: iterate models and splits
model_files = sorted([p for p in models_dir.iterdir() if p.suffix == '.pt'])
if not model_files:
    raise RuntimeError(f'No .pt models found in {models_dir}')

splits = ['val', 'test']  # user asked for both validation and test; our folders are 'valid' and 'test'
# map 'val' to 'valid' internally in evaluate function

for m in model_files:
    for s in splits:
        evaluate_model_on_split(m, s)

print('\nAll evaluations finished. Check the evaluation_outputs_of_1008 folder for results.')
```

```python
# Concise parser: prints per-epoch train/val losses from Ultralytics metrics CSV for YOLOv8s
import os, glob, csv

def print_epoch_losses(run_dir):
    run_dir = os.path.normpath(run_dir)
    candidates = [os.path.join(run_dir, 'metrics.csv'), os.path.join(run_dir, 'results.csv')] + glob.glob(os.path.join(run_dir, '*.csv'))
    csv_path = next((p for p in candidates if os.path.exists(p)), None)
    if not csv_path:
        print(f'No CSV metrics found in {run_dir}')
        return
    with open(csv_path, newline='') as f:
        reader = csv.DictReader(f)
        rows = list(reader)
        hdrs = reader.fieldnames or []
    if not rows:
        print(f'Found CSV at {csv_path} but it has no rows')
        return
    # common column name heuristics (kept minimal)
    epoch_col = next((h for h in hdrs if h.lower() in ('epoch', 'ep')), None)
    train_col = next((h for h in hdrs if 'train' in h.lower() and 'loss' in h.lower()), None) or next((h for h in hdrs if h.lower() == 'loss'), None)
    val_col = next((h for h in hdrs if ('val' in h.lower() or 'valid' in h.lower()) and 'loss' in h.lower()), None) or next((h for h in hdrs if 'val' in h.lower()), None)

    for i, row in enumerate(rows, start=1):
        e = row.get(epoch_col) if epoch_col else i
        try:
            e = int(float(e))
        except Exception:
            e = i
        train_loss = float(row.get(train_col, 0.0)) if train_col else 0.0
        valid_loss = float(row.get(val_col, 0.0)) if val_col else 0.0
        print('Epoch: {} \tTraining Loss: {:.6f} \tValidation Loss: {:.6f}'.format(e, train_loss, valid_loss))

# Usage: change the path below to your run folder if needed
print_epoch_losses('model_outputsof_1008_images/yolov8s')
```

```python
# Plot training and validation loss vs epoch from Ultralytics metrics CSV
import os, glob, csv
import matplotlib.pyplot as plt


def plot_epoch_losses(run_dir, show=True, save_path=None):
    run_dir = os.path.normpath(run_dir)
    candidates = [os.path.join(run_dir, 'metrics.csv'), os.path.join(run_dir, 'results.csv')] + glob.glob(os.path.join(run_dir, '*.csv'))
    csv_path = next((p for p in candidates if os.path.exists(p)), None)
    if not csv_path:
        print(f'No CSV metrics found in {run_dir}')
        return
    epochs, train_losses, val_losses = [], [], []
    with open(csv_path, newline='') as f:
        reader = csv.DictReader(f)
        hdrs = reader.fieldnames or []
        epoch_col = next((h for h in hdrs if h.lower() in ('epoch', 'ep')), None)
        train_col = next((h for h in hdrs if 'train' in h.lower() and 'loss' in h.lower()), None) or next((h for h in hdrs if h.lower() == 'loss'), None)
        val_col = next((h for h in hdrs if ('val' in h.lower() or 'valid' in h.lower()) and 'loss' in h.lower()), None) or next((h for h in hdrs if 'val' in h.lower()), None)
        for i, row in enumerate(reader, start=1):
            e = row.get(epoch_col) if epoch_col else i
            try:
                e = int(float(e))
            except Exception:
                e = i
            epochs.append(e)
            train_losses.append(float(row.get(train_col, 0.0)) if train_col else 0.0)
            val_losses.append(float(row.get(val_col, 0.0)) if val_col else 0.0)

    plt.figure(figsize=(8,5))
    plt.plot(epochs, train_losses, label='Train Loss', marker='o')
    plt.plot(epochs, val_losses, label='Val Loss', marker='o')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Training and Validation Loss')
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    if save_path:
        plt.savefig(save_path)
        print(f'Plot saved to {save_path}')
    if show:
        plt.show()


# Usage: change run_dir if needed
plot_epoch_losses('model_outputsof_1008_images/yolov8s')
```

The training loss consistently decreases, indicating that the model is effectively learning the patterns in the training data and improving its performance on that data. The validation loss also decreases initially, but it fluctuates significantly and is consistently higher than the training loss after the first few epochs

The gap between the training loss and validation loss, known as the "generalization gap," is a key indicator of model performance. The fact that the training loss continues to decrease while the validation loss is consistently higher and fluctuates indicates that the model may be overfitting

Overfitting occurs when a model learns the training data too well, including noise and specific details, and loses its ability to generalize to new, unseen data. While a small generalization gap is normal, a large and increasing gap suggests that the model is "memorizing" the training data rather than learning general features. 
The validation loss has a noisy or erratic pattern, which could be a sign of issues like a high learning rate or noisy data. The training loss, however, is much smoother and more consistent

Areas for Improvement
To address the potential overfitting and improve the model's performance on unseen data, the following improvements could be necessary:
Regularization: The model could be simplified or regularized to reduce its complexity. Techniques such as adding dropout layers or using L2 normalization can help prevent the model from becoming too reliant on the training data.
Early Stopping: Since the validation loss stops improving and fluctuates after a certain point, a technique called "early stopping" could be used to halt the training process before the model begins to overfit. This would involve stopping training when the validation loss stops decreasing for a specified number of epochs.
Data Augmentation: Increasing the amount of training data or using data augmentation techniques can help the model learn more robust patterns and improve its ability to generalize.
Learning Rate Adjustment: An erratic or unstable learning curve can be caused by a high learning rate. Reducing the learning rate or using an adaptive learning rate scheduler might lead to a smoother and more stable training process

```python
# Plot training and validation loss vs epoch from Ultralytics metrics CSV
import os, glob, csv
import matplotlib.pyplot as plt


def plot_epoch_losses(run_dir, show=True, save_path=None):
    run_dir = os.path.normpath(run_dir)
    candidates = [os.path.join(run_dir, 'metrics.csv'), os.path.join(run_dir, 'results.csv')] + glob.glob(os.path.join(run_dir, '*.csv'))
    csv_path = next((p for p in candidates if os.path.exists(p)), None)
    if not csv_path:
        print(f'No CSV metrics found in {run_dir}')
        return
    epochs, train_losses, val_losses = [], [], []
    with open(csv_path, newline='') as f:
        reader = csv.DictReader(f)
        hdrs = reader.fieldnames or []
        epoch_col = next((h for h in hdrs if h.lower() in ('epoch', 'ep')), None)
        train_col = next((h for h in hdrs if 'train' in h.lower() and 'loss' in h.lower()), None) or next((h for h in hdrs if h.lower() == 'loss'), None)
        val_col = next((h for h in hdrs if ('val' in h.lower() or 'valid' in h.lower()) and 'loss' in h.lower()), None) or next((h for h in hdrs if 'val' in h.lower()), None)
        for i, row in enumerate(reader, start=1):
            e = row.get(epoch_col) if epoch_col else i
            try:
                e = int(float(e))
            except Exception:
                e = i
            epochs.append(e)
            train_losses.append(float(row.get(train_col, 0.0)) if train_col else 0.0)
            val_losses.append(float(row.get(val_col, 0.0)) if val_col else 0.0)

    plt.figure(figsize=(8,5))
    plt.plot(epochs, train_losses, label='Train Loss', marker='o')
    plt.plot(epochs, val_losses, label='Val Loss', marker='o')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Training and Validation Loss')
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    if save_path:
        plt.savefig(save_path)
        print(f'Plot saved to {save_path}')
    if show:
        plt.show()


# Usage: change run_dir if needed
plot_epoch_losses('model_outputsof_1008_images/yolov8m')
```

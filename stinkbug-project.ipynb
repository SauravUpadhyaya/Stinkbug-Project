```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import os
import io
from PIL import Image

def draw_boxes_on_image(image_path, label_path):
    img = Image.open(image_path).convert("RGB")
    img_width, img_height = img.size

    fig, ax = plt.subplots(1)
    ax.imshow(img)

    if os.path.exists(label_path):
        with open(label_path, 'r') as f:
            for line in f.readlines():
                parts = line.strip().split()
                if len(parts) == 5:
                    class_id, x_center, y_center, width, height = map(float, parts)
                    x1 = (x_center - width / 2) * img_width
                    y1 = (y_center - height / 2) * img_height
                    box_width = width * img_width
                    box_height = height * img_height
                    rect = patches.Rectangle((x1, y1), box_width, box_height, linewidth=1, edgecolor='r', facecolor='none')
                    ax.add_patch(rect)


    ax.set_axis_off()
    plt.tight_layout()
    buf = io.BytesIO()
    fig.savefig(buf, format='png')
    buf.seek(0)
    annotated_img = Image.open(buf).convert("RGB")
    plt.close(fig)

    return annotated_img

for img_path, label_path in image_label_pairs[:5]:
    annotated_img = draw_boxes_on_image(img_path, label_path)
    fig, ax = plt.subplots(figsize=(6, 6))
    ax.imshow(annotated_img)
    ax.set_title(os.path.basename(img_path))
    ax.axis('off')
    plt.show()
```

```python
import os
import csv
from ultralytics import YOLO
import cv2


model_dir = "saved_models"          
test_images_dir = "./stinkbug_image_data/test/images"    
output_root = "model_predictions"  
metrics_csv = "object_counts.csv"   

os.makedirs(output_root, exist_ok=True)

# Collect all model files
model_files = [f for f in os.listdir(model_dir) if f.endswith(".pt")]
image_files = [f for f in os.listdir(test_images_dir) if f.lower().endswith((".jpg", ".png", ".jpeg"))]

# Open CSV to log object counts
with open(metrics_csv, mode="w", newline="") as csvfile:
    writer = csv.writer(csvfile)
    header = ["image"] + [model_file.replace("_best.pt", "") for model_file in model_files]
    writer.writerow(header)

    for image_file in image_files:
        image_path = os.path.join(test_images_dir, image_file)
        row = [image_file]

        for model_file in model_files:
            model_name = model_file.replace("_best.pt", "")
            model_path = os.path.join(model_dir, model_file)
            model = YOLO(model_path)

            # Predict
            results = model.predict(source=image_path, conf=0.25, save=False, imgsz=640)

            # Count number of detections
            preds = results[0].boxes
            num_objects = len(preds)
            row.append(num_objects)

            # Save visualization
            result_img = results[0].plot() 

            # Create output dir for this model
            model_output_dir = os.path.join(output_root, model_name)
            os.makedirs(model_output_dir, exist_ok=True)

            # Save image
            save_path = os.path.join(model_output_dir, image_file)
            cv2.imwrite(save_path, result_img)

            print(f"Saved: {save_path} | Count: {num_objects} objects")

        writer.writerow(row)

print(f"\n All done! Object counts saved to: {metrics_csv}")
```

```python
def get_counts_from_labels(labels_dir, csv_output_path):
    import os
    import csv

    counts = {}
    for label_file in os.listdir(labels_dir):
        if label_file.endswith(".txt"):
            with open(os.path.join(labels_dir, label_file)) as f:
                counts[label_file.replace('.txt', '.jpg')] = len(f.readlines())
    
    with open(csv_output_path, 'w', newline='') as csvfile:
        writer = csv.writer(csvfile)
        writer.writerow(['image', 'true_count'])
        for img, count in counts.items():
            writer.writerow([img, count])
    
    print(f"Saved ground truth counts to {csv_output_path}")

# Usage:
get_counts_from_labels("./stinkbug_image_data/train/labels", "train_ground_truth_counts.csv")
get_counts_from_labels("./stinkbug_image_data/valid/labels", "val_ground_truth_counts.csv")
get_counts_from_labels("./stinkbug_image_data/test/labels", "test_ground_truth_counts.csv")
```

```python
import numpy as np

def calculate_metrics(true_counts, pred_counts):
    N = len(true_counts)
    min_vals = np.minimum(true_counts, pred_counts)
    max_vals = np.maximum(true_counts, pred_counts)
    aca = np.sum(min_vals / max_vals) / N

    detection_rate = np.sum(pred_counts >= true_counts) / N
    failed_detection_rate = 1 - detection_rate

    return aca, detection_rate, failed_detection_rate

results = []
for model in pred_counts.columns[1:]: 
    aca, dr, fdr = calculate_metrics(df['true_count'], df[model])
    results.append({
        'model': model,
        'Average Counting Accuracy': aca,
        'Detection Rate': dr,
        'Failed Detection Rate': fdr
    })

metrics_df = pd.DataFrame(results)
print(metrics_df)
metrics_df.to_csv("model_counting_metrics.csv", index=False)
```

```python
import pandas as pd
import numpy as np

def calculate_counting_metrics(true_counts, pred_counts):
    """Compute ACA, MAE, DR, and FDR between predicted and true counts."""
    true_counts = np.array(true_counts)
    pred_counts = np.array(pred_counts)

    # Avoid divide by zero
    max_vals = np.maximum(true_counts, pred_counts)
    aca = np.mean(np.where(max_vals == 0, 1, np.minimum(true_counts, pred_counts) / max_vals))
    mae = np.mean(np.abs(true_counts - pred_counts))
    dr = np.mean(pred_counts >= true_counts)
    fdr = 1 - dr

    return round(aca, 3), round(mae, 3), round(dr, 3), round(fdr, 3)

# === Load your data ===

ground_truth_csv = "test_ground_truth_counts.csv"   # columns: image,true_count
predictions_csv = "object_counts.csv"            # columns: image, yolov8n, yolov8s, ...

# Load
gt_df = pd.read_csv(ground_truth_csv)
pred_df = pd.read_csv(predictions_csv)

# Merge on image
df = pd.merge(gt_df, pred_df, on="image")

# === Evaluate models ===

metrics = []
for model in pred_df.columns[1:]: 
    aca, mae, dr, fdr = calculate_counting_metrics(df["true_count"], df[model])
    metrics.append({
        "Model": model,
        "Average Counting Accuracy": aca,
        "Mean Absolute Error": mae,
        "Detection Rate": dr,
        "Failed Detection Rate": fdr
    })

metrics_df = pd.DataFrame(metrics)
metrics_df.to_csv("model_counting_metrics.csv", index=False)
print(metrics_df)
```
